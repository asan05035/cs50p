# Passing functions inside another function
    - two types
    - 1. Hand you back some value ...fn()
    - 2. function to be used later ... fn

```python
def main():
    print(calculate_square(get_int(), get_square))

def get_int():
    return int(input("Enter x: "))

def get_square(n):
    return n ** 2

def calculate_square(n, square):
    return square(n)

if __name__ == "__main__":
    main()
```
