# File I/O

# 0. Loss of information during execution

```python
names = []

for _ in range(0, 3):
    names.append(input("What's your name?: "))

for name in sorted(names): #Loop kicks in
    print(f"hello, {name}")
```

# 1. Writing into file
    - write mode (overwrite)
    - append mode

```python
name = input("What's your name?: ")

with open("names.txt", "a") as file:
    file.write(f"{name}\n")
```

# 2. Reading the file

```python
with open("names.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(f"Hello, {line}")
    
```

## 2.1 Reading entire file and iterate each line

```python
with open("names.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(f"Hello, {line.rstrip()}")
```

## 2.2 Reading each line (one at a time)

```python
with open("names.txt", "r") as file:
    for line in file:
        print(f"Hello, {line.rstrip()}")
```

# 3. Accumulation of information

```python
names = []

with open("names.txt") as file:
    for line in file:
        names.append(line.rstrip())

for name in sorted(names, reverse=True):
    print(f"Hello, {name}")
```

# 4. comma separated value 
    - an0ther type of text file to store information
    - avoid mixing of info(apple and oranges)

```python
with open("students.csv") as file:
    for line in file:
        row = line.rstrip().split(",")
        print(f"{row[0]} is in {row[1]}")   #Cryptic
```

# 5. List of dictionary
    -   Readability and not to make things llok cryptic

```python
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)

for student in students:
    print(f"{student['name']} is in {student['house']}")
        
```
