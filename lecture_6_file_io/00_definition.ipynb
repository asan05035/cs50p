# 1. Writing into file
    - write mode (overwrite)
    - append mode

```python
name = input("What's your name?: ")

with open("names.txt", "a") as file:
    file.write(f"{name}\n")
```

# 2. Reading the file

```python
with open("names.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(f"Hello, {line}")
    
```

## 2.1 Reading entire file and iterate each line

```python
with open("names.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(f"Hello, {line.rstrip()}")
```

## 2.2 Reading each line (one at a time)

```python
with open("names.txt", "r") as file:
    for line in file:
        print(f"Hello, {line.rstrip()}")
```

# 3. Accumulation of information

```python
names = []

with open("names.txt") as file:
    for line in file:
        names.append(line.rstrip())

for name in sorted(names, reverse=True):
    print(f"Hello, {name}")
```

# 5. List of dictionary
    -   Readability and not to make things llok cryptic

```python
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)

for student in students:
    print(f"{student['name']} is in {student['house']}")
        
```

# 5. sort based on keys

```python
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)

def get_name(student):
    return student["name"]

for student in sorted(students, key=get_name, reverse=True):
    print(f"{student['name']} is in {student['house']}")
```

# 6.lambda funcitons
    - Anoymous function( no_name function)
    - define function -> using it -> (not needed again vs reusablilty)

```python
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)


for student in sorted(students, key=lambda student: student["name"], reverse=True):
    print(f"{student['name']} is in {student['house']}")
```

# 6.1 Lambda funciton usage example
    - Passing function to function

```python
def square(n):
    return n ** 2

def cube(n):
    return n ** 3

def calculate(n, function):
    return function(n)


number = 3
print(f"Suare of 3 is: {calculate(number, square)}")
print(f"cube of 3 is: {calculate(number, cube)}")
```

```python
def calculate(n, function):
    return function(n)


number = 3
print(f"Suare of 3 is: {calculate(number, lambda n : n ** 2)}")
print(f"cube of 3 is: {calculate(number, lambda n : n ** 3)}")
```
