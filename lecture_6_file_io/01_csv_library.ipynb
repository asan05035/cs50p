# 1.1 specify only on comma outside quotes

```python
students = []

with open("students.csv") as file:
    for line in file:
        name, house = line.rstrip().split(",")
        student = {"name": name, "house": house}
        students.append(student)


for student in sorted(students, key=lambda student: student["name"], reverse=True):
    print(f"{student['name']} is in {student['house']}")
```

# 1.2 csv.Reader
    - I'll handle the corner case for you

```python
import csv 
students = []

with open("students.csv") as file:
    reader = csv.reader(file)
    for name, home in reader:
        stduent = {"name": name, "home": home}
        students.append(stduent)

for student in sorted(students, key=lambda student:student["name"]):
    print(f"{student['name']} from {student['home']}")
```

# 1.3 csv DictReader

```python
import csv 
students = []

with open("students.csv") as file:
    reader = csv.DictReader(file)
    for row in reader:
        stduent = {"name": row["name"], "home": row["home"]}
        students.append(stduent)

for student in sorted(students, key=lambda student:student["name"]):
    print(f"{student['name']} from {student['home']}")
```

# 1.4 csv writer

```python
name = input("What's your name: ")
home = input("where did you live: ")

with open("students.csv", "a") as outfile:
    writer = csv.writer(outfile)
    writer.writerow([name, home])
```
