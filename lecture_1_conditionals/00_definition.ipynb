# 1. Conditionals
    - way of asking questions
    - putting proverbial fork in your code

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x > y:
    print("x is greater than y")
if x < y:
    print("x is less than y")
if x == y:
    print("x is equal to y")
```

# 2. Instead of asking question continously it is beeter to ask till u got a True answer

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x > y:
    print("x is greater than y")
elif x < y:
    print("x is less than y")
else:
    print("x is equal to y")
```

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x != y:
    print("x is not equal to y")
else:
    print("x is equal to y")
```

# 3. logical opeartors and chaining questions

```python
score: int = int(input("Score: "))

if 90 <= score <= 100:
    print("Grade A")
elif 80 <= score < 90:
    print("Grade B")
else:
    print("Grade E")
```
