# 1. Conditionals
    - way of asking questions
    - putting proverbial fork in your code

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x > y:
    print("x is greater than y")
if x < y:
    print("x is less than y")
if x == y:
    print("x is equal to y")
```

# 2. Instead of asking question continously it is beeter to ask till u got a True answer

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x > y:
    print("x is greater than y")
elif x < y:
    print("x is less than y")
else:
    print("x is equal to y")
```

```python
x: int = int(input("What's x? "))
y: int = int(input("What's y? "))

if x != y:
    print("x is not equal to y")
else:
    print("x is equal to y")
```

# 3. logical opeartors and chaining questions

```python
score: int = int(input("Score: "))

if 90 <= score <= 100:
    print("Grade A")
elif 80 <= score < 90:
    print("Grade B")
else:
    print("Grade E")
```

# 5. boolean expressions :
    - something thats True or False

```python
x: int = int(input("What's x? "))

def main() -> None:
    if is_even(x):              # Naively aassume i have the ability
        print("Even")
    else:
        print("Odd")


def is_even(n: int) -> bool:    # Implement it
    return n % 2 == 0


if __name__ == "__main__":
    main()
```

# match case
    - spritually eqial to switch in other programming language

```python
name: str = input("What's your name? ")

match name:
    case "Harry" | "Hermoine" | "Ron":
        print("Gryyfindor")
    case "Draco":
        print("Slytherin")
    case _:             # if all the above cases fails then default path
        print("Who? ")
```
