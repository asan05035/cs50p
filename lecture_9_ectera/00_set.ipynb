# Set
    - auto Filerring of duplicates

```python
students = [
            {"name": "harry", "house": "Gryfinnfdor"},
            {"name": "ron", "house": "Gryfinnfdor"},
            {"name": "padma", "house": "Ravenclaw"},
            {"name": "draco", "house": "Slytherin"},
]


houses = []  # Accumaltion of information
for student in students:
    if student["house"] not in houses:  # Filtering
        houses.append(student["house"])

for house in sorted(houses):
    print(f"{house}")

```

# using set 

```python
students = [
            {"name": "harry", "house": "Gryfinnfdor"},
            {"name": "ron", "house": "Gryfinnfdor"},
            {"name": "padma", "house": "Ravenclaw"},
            {"name": "draco", "house": "Slytherin"},
]


houses = set()  # Accumaltion of information    # Take care of filtering
for student in students:
    houses.add(student["house"])    

for house in sorted(houses):
    print(f"{house}")
```
