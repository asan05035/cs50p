# Unpacking vs packing
    - container with name or not

```python
def total(galleons: int, sickles: int, knuts: int) -> int:
    """
    Calculating total knuts by converting galleon and sickles.
    """
    return galleons * 479 + sickles * 29 + knuts

coins: list[int] = [20, 100, 300]

print(f"{total(*coins)} knuts")
```

```python
coins: dict[str, int] = {"galleons": 20, "sickles": 100, "knuts": 300}
print(total(**coins), "knuts")
```
