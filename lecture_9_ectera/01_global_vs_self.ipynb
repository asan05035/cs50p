# With global key
    - we got access to write 

```python
balance = 0


def main():
    print("Balance: ", balance)
    deposit(100)
    withdrawl(50)
    print("Balance", balance)


def deposit(n):
    global balance
    balance += n


def withdrawl(n):
    global balance
    balance -= n


if __name__ == "__main__":
    main()
```

# Encapsultaion

```python
class Account:
    def __init__(self):
        self._balance = 0

    @property
    def balance(self):
        return self._balance


    def deposit(self, n):
        self._balance += n


    def withdrawl(self, n):
        self._balance -= n



def main():
    account = Account()
    print(f"Balance: {account.balance}")
    account.deposit(100)
    account.withdrawl(50)
    print(f"Balance: {account.balance}")


if __name__ == "__main__":
    main()
```
