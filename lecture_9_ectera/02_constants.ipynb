# Constant
    - All caps to indicate it is constant, dont touch it

```python
for i in range(3):
    print("meow")
```

```python
MEOW = 3

for _ in range(MEOW):
    print("meow")
```

```python
class Cat:
    MEOW = 3

    def meow(self):
        for _ in range(Cat.MEOW):
            print("meow")

cat = Cat()
cat.meow()
```
