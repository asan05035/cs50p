# map
    - functional prograaming

```python
def main():
    yell("This",  "is",  "CS50")


def yell(*words):
    uppercased = []     #Accumulation of information
    for word in words:
        uppercased.append(word.upper())
    print(*uppercased)  # *objects - unpacking


if __name__ == "__main__":
    main()
    
```

# map function 
    - whose purpose in its life apply some function to each item in the sequence

```python
def main():
    yell("This",  "is",  "CS50")


def yell(*words):
    uppercased = map(str.upper, words)  # passing funtion to another function
    print(*uppercased)

if __name__ == "__main__":
    main()
    
```
