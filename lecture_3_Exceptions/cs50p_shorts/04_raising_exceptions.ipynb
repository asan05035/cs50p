```python
def main():
    pace = get_pace(miles=182.6, minutes=0)
    print(f"You need to run {pace} minutes per mile")

def get_pace(miles, minutes):
    if not minutes > 0: #Flip the shade of numerical scal to left side
        raise ValueError("Minutes must be greater than zero")
    return minutes/miles

if __name__ == "__main__":
    main()
```
