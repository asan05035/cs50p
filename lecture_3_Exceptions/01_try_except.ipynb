# 2. Else block will only run if the try block runs sucessfully
    - nice to have skill

```python
try:
    x = int(input("What's x "))

except ValueError:
    print(f'x is a string, it has to be an integer')

else:
    print(f'x ix {x}')
```

# 3. Prompting...break
    - if u enter a loop, u use break to get out of the loop

```python
while True:
    try:
        x = int(input("What's x "))

    except ValueError:
        print(f'x is a string, it has to be an integer')

    else:
        break

print(f'x is {x}')
```

# 4. Inventing our own fuction
    - whose purpose in its life 
        - print: Something on the screen has side effects (sfx)
        - return: Hand you back some value

```python
def main():
    x = get_int()
    print(f"x is {x}")

def get_int():
    while True:
        try:
            x = int(input("What's x "))

        except ValueError:
            print(f'x is a string, it has to be an integer')

        else:
            break
    return x


if __name__ == '__main__':
    main()
```
