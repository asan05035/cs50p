```python
import requests
import sys
import json
```

```python
artist_name = input("Artist name: ")

if not artist_name:
    sys.exit()
```

```python
response = requests.get('https://itunes.apple.com/search?limit=20&entity=song&term=' + artist_name)
print(json.dumps(response.json(), indent=2))
```

```python
o = response.json()
for result in o["results"]:
    print(result["trackName"])
```
