# 1 Function Types  (eg. randint, shuffle)
    - takes input , process it, produces output
    - modifies the input

```python
from random import shuffle, randint

# Using randint - random number from a to b
random_num = randint(1, 10)
print(random_num)

print('-' * 30)
# Using shuffle to shuffle things
cards = ["king", "queen", "jack"]
shuffle(cards)
for card in cards:
    print(card)
```
