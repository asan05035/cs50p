# 0. Regex 
    - pattern serach
    - pattern building

```python
email = input("What's your email? ").strip()
```

# without regex

```python
username, domain = email.split("@")

if username and "." in domain:
    print("Valid")
else:
    print("Invalid")
```

# with regex

```python
import re
```

# 1.1 re.search(patter, search, flags = 0)
    - pattern to serach for
    - search against or in string

```python
if re.search("@", email):
    print("Valid")
else:
    print('Invalid')
```
