# 3.1 Class method

```python
class Hat:
    def sort(self, name):
        print(name, "is in", "some house")

hat = Hat()
hat.sort("Harry")
```

# 3.2 Convention in intiate 

```python
import random

class Hat:
    def __init__(self):
        self.houses = ["Gryyfindor", "Hufflepuff", "Ravenclaw", "Slytherin"]
    
    def sort(self, name):
        house = random.choice(self.houses)
        print(name, "is in", house)

hat = Hat()
hat.sort("Harry")
```

# 3.2 Class has a container (Not has blueprint)

```python
import random

class Hat:
    houses = ["Gryyfindor", "Hufflepuff", "Ravenclaw", "Slytherin"]

    @classmethod
    def sort(cls, name):
        print(f'{name} is in {random.choice(cls.houses)}')

Hat.sort("Ron")
```
