# 1.1 Property vs attribute
    - Property is fancier attribute with extra defense mechanism in it
    - If i go through the trouble making validation, i dont want someone to circumvent that


```python
class Student:
    def __init__(self, name, house):
        # Getting inside the otherwise empty object and putting inside the details
        if not name:
            # Helping python in catching error by giving precise message
            raise ValueError('Missing Name')
        if house not in ['Gryyfindor', 'Huffelpuff', 'Ravenclaw','Slytherin' ]:
            raise ValueError('Invalid House')   #Like function we can pass message in error
        self.name = name
        self.house = house

    # String method
    def __str__(self):
        return f'{self.name} from {self.house}'


def main():
    student = get_student()
    student.name = 'Padma'  
    print(f'Circumventing the validation')
    print(student)


def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    # Tightening things by directly returning the object 
    # instead storing it to a varibale
    return Student(name, house)  #Calling the constructor    
    #Returning an object

if __name__ == '__main__':
    main()  
```

# 1.2 Preventing direct access to the attribute
    - getter function is a function inside class to get the attribute value
    - setter function is used to set the value

```python
class Student:
    def __init__(self, name, house):
        # Getting inside the otherwise empty object and putting inside the details
        if not name:
            # Helping python in catching error by giving precise message
            raise ValueError('Missing Name')
         
        self.name = name
        self.house = house

    # String method
    def __str__(self):
        return f'{self.name} from {self.house}'

    #Getter
    @property 
    def house(self):
        return self._house

    #Setter
    @house.setter
    def house(self, house):
        if house not in ['Gryyfindor', 'Huffelpuff', 'Ravenclaw','Slytherin' ]:
            raise ValueError('Invalid House') #Like function we can pass message in error
        self._house = house

def main():
    student = get_student()
    #student.house = 'Privete Four, Direct house'  
    #print(f'Circumventing the validation')
    print(student)


def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    # Tightening things by directly returning the object 
    # instead storing it to a varibale
    return Student(name, house)  #Calling the constructor    
    #Returning an object

if __name__ == '__main__':
    main()  
```
