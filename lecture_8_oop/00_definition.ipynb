# 1.1 List vs Tuple
    - Use tuple when u want values stored in the variable to not change
    - () visual indicator for tuple


```python
def main():
    student = get_student()
    print(f'{student[0]} from {student[1]}')

def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    return name, house      #Mere fact of using comma --> python understand it is tuple

if __name__ == '__main__':
    main()
```

# 1.2 Dictionary
    - upside is sematic asscociation(he --> baara, she --> samara)
    - Dictionary is mutable like list but keyed insted of indexed

```python
def main():
    student = get_student()
    if student['name'] == 'Padma':
        student['house'] = 'Ravenclaw' # Dictionary is mutable like list
    print(f'{student["name"]} from {student["house"]}')

def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    return {'name' : name, 'house' :house }     

if __name__ == '__main__':
    main()
```

```python
class Student:
    def __init__(self, name, house):
        # Getting inside the otherwise empty object and putting inside the details
        self.name = name
        self.house = house

def main():
    student = get_student()
    print(f'{student.name} from {student.house}')

def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    # Tightening things by directly returning the object 
    # instead storing it to a varibale
    return Student(name, house)  #Calling the constructor    
    #Returning an object

if __name__ == '__main__':
    main()  
```

# 1.7 Custom methods

```python
class Student:
    def __init__(self, name, house, patronus):
        # Getting inside the otherwise empty object and putting inside the details
        if not name:
            # Helping python in catching error by giving precise message
            raise ValueError('Missing Name')
        if house not in ['Gryyfindor', 'Huffelpuff', 'Ravenclaw','Slytherin' ]:
            raise ValueError('Invalid House')   #Like function we can pass message in error
        self.name = name
        self.house = house
        self.patronus = patronus

    # String method
    def __str__(self):
        return f'{self.name} from {self.house}'


    # Custome method
    def charm(self):
        match self.patronus:
            case 'stag':
                return '🦌'
            case 'Otter':
                return '🦦'
            case _:
                return '🪄'
        
def main():
    student = get_student()
    print(student)
    print('Expecto patrnous!!!!')
    print(student.charm())

def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    patronus = input('Patronus: ')
    # Tightening things by directly returning the object 
    # instead storing it to a varibale
    return Student(name, house, patronus)  #Calling the constructor    
    #Returning an object

if __name__ == '__main__':
    main()  
```
