# 4.1 Operator overloading
    - specail method
    - redefine how builtin oprator behave on my own defined objects

```python
class Vault:
    def __init__(self, galleons=0, sickle=0, knuts=0):
        self.galleons = galleons
        self.sickles = sickle
        self.knuts = knuts

    def __str__(self):
        return f"Galleons: {self.galleons}, Sickles: {self.sickles}, Knuts: {self.knuts}"
```

```python
harry = Vault(125, 50, 100)
print(harry)
```

```python
ron = Vault(10, 10, 50)
print(ron)
```

```python
harry + ron
```

# 4.2 Way_1: To avoid operand typeError

```python
total_galleon = harry.galleons + ron.galleons
total_sickle = harry.sickles + ron.sickles
total_knuts = harry.knuts = ron.knuts

total_vault = Vault(total_galleon, total_sickle, total_knuts)
print(total_vault)
```

# 4.2 Pythonic way
    - Teaching python using special method
    - object.__add__(self, other)
    - operands, operator

```python
class Vault:
    def __init__(self, galleons=0, sickle=0, knuts=0):
        self.galleons = galleons
        self.sickles = sickle
        self.knuts = knuts

    def __str__(self):
        return f"Galleons: {self.galleons}, Sickles: {self.sickles}, Knuts: {self.knuts}"

    def __add__(self, other):
        galleons = self.galleons + other.galleons # Local variable
        sickles = self.sickles + other.sickles
        knuts = self.knuts + other.knuts
        return Vault(galleons, sickles, knuts)
```

```python
harry = Vault(120, 50, 100)
ron = Vault(20, 30, 10)
```

```python
total = harry + ron
print(total)
```
