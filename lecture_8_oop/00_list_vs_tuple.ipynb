# List vs Tuple
    - Use tuple when u want values stored in the variable to not change
    - () visual indicator for tuple


```python
def main():
    student = get_student()
    print(f'{student[0]} from {student[1]}')

def get_student():
    name = input('Name: ') #Prompting the user
    house = input('House: ')
    return name, house      #Mere fact of using comma --> python understand it is tuple

if __name__ == '__main__':
    main()
```
